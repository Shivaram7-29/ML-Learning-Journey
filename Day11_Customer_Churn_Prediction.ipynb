```python
import pandas as pd 
df = pd.read_csv("/kaggle/input/datasets/blastchar/telco-customer-churn/WA_Fn-UseC_-Telco-Customer-Churn.csv")
df.head()
```

**Basic Cleaning**

```python
df = df.drop(columns=["customerID"],errors='ignore')
df = df[["gender", "tenure", "MonthlyCharges", "TotalCharges", "Churn"]]#only keep numeric columns for safe as beginer
## To Avoid NaN in output use the code below
df["gender"] = df["gender"].str.strip().str.lower()
df["Churn"] = df["Churn"].str.strip().str.lower()
df["TotalCharges"] = df["TotalCharges"].str.strip().str.lower()
df["gender"] = df["gender"].map({"male":1,"female":0})
df["Churn"] = df["Churn"].map({"yes":1,"no":0}) ## yes--> left , No --> stay
df["TotalCharges"] = pd.to_numeric(df["TotalCharges"], errors='coerce')
df["TotalCharges"] = df["TotalCharges"].fillna(df["TotalCharges"].mean())
df = df.fillna(df.mean(numeric_only=True))
df.head()
```

**Features & Target**

```python
X = df.drop("Churn",axis=1) # this says: drop the churn column and keep everything as input
y = df["Churn"]  # target
```

**Train-Test Split**

```python
from sklearn.model_selection import train_test_split

X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2)
## You are doing:

# Give some data to learn
# Keep some data hidden for testing
```

**Train Model**

```python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression()
model.fit(X_train, y_train)
# Model looks at your data and learns patterns like:

# If MonthlyCharges is high → maybe churn
# If tenure is low → maybe churn

# 📊 WHAT ARE YOU GIVING TO MODEL?
# model.fit(X_train, y_train)
# You give:
# X_train → input (features)
# y_train → correct answers 

#WHY Logestic regression 
# Your task is:
# Predict Churn
# Output is:
# 0 → No
# 1 → Yes
# 👉 This is called:
# 👉 Binary Classification (2 classes)

# 🎯 WHY LOGISTIC REGRESSION?
# 👉 Because it is made for:
# ✔ Yes / No problems
# ✔ 0 / 1 prediction
# ✔ Simple classification
```

**PREDICT**

```python
pred = model.predict(X_test)
#X_test (new/unseen data)
print(pred)
```

**Accuracy**

```python
from sklearn.metrics import accuracy_score

accuracy = accuracy_score(y_test, pred)
print(accuracy)
# you are comparing:

# Actual answers → y_test
# Model answers → pred
```
