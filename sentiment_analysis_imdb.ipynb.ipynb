**Step 3: Clean Text (BASIC)**

```python
df["review"] = df["review"].str.lower()
df["review"] = df["review"].str.replace(r"<.*?>", "", regex=True)
# 🧠 1. What is “cleaning text”?

# 👉 Your raw data looks like this:
# "A wonderful little production. <br /><br />The acting was GREAT!"
# Problems:
# Uppercase (GREAT)
# HTML tags (<br />)
# Extra symbols / noise

# 👉 ML model gets confused by this.

# Remove HTML tags:
# Your data has:
# <br /><br />  👉 This is useless for sentiment
# So we remove it:

# df["review"] = df["review"].str.replace(r"<.*?>", "", regex=True)
# 🧠 What this regex means (simple)
# <.*?>

# < → start of tag
# .*? → anything inside
# > → end of tag

# 👉 So it removes:

# <br>  <br />  <div> etc
```

**Step 4: Convert Text → Numbers (IMPORTANT)**

```python
from sklearn.feature_extraction.text import CountVectorizer

cv = CountVectorizer()
X = cv.fit_transform(df["review"])
# 🧠 Core Problem

# Right now your data looks like:

# "a wonderful movie"
# "bad acting and boring"


# 👉 ML model CANNOT understand words
# 👉 It only understands:
# [1, 0, 3, 5, ...]

# So we use CountVectorizer

# 🧠 What is CountVectorizer?
# 👉 It uses a concept called:

# 👉 “Bag of Words”

# 🎒 Bag of Words (simple idea)
# 👉 Take all reviews → collect all unique words

# Example:

# Review 1: "good movie"
# Review 2: "bad movie"

# 👉 Vocabulary becomes:

# ["good", "movie", "bad"]

# Now convert each sentence
# Review 1: "good movie"
# good → 1
# movie → 1
# bad → 0

# → [1, 1, 0]

# Review 2: "bad movie"
# good → 0
# movie → 1
# bad → 1

# → [0, 1, 1]

# 👉 THIS is what CountVectorizer does automatically
```

🔍 What your code is doing
**cv = CountVectorizer()**
👉 Creates the vectorizer (tool)

**X = cv.fit_transform(df["review"])**

👉 Two things happen:

1. fit()
Learns all unique words (vocabulary)
2. transform()
Converts each review → numeric vector

🧠 What is X?
👉 X is NOT normal data

It is a:

👉 Sparse Matrix (memory efficient)
🔍 See it yourself

Run:
print(X.shape)
👉 Output:

(50000, 5000+)   # approx

Meaning:

5000+ unique words

```python
print(cv.get_feature_names_out())
#👉 Shows all words learned
# 🔍 Your output:
# ['00' '000' '00000000000' ... 'żmijewski' 'יגאל' 'כרמון']

# 🔥 What this means
# ✔ Numbers like:
# '00', '000'

# 👉 These came from reviews like:

# "10/10 movie", "100% great"

# ✔ Weird long strings:
# '00000000000'

# 👉 Probably noise / repeated characters

# ✔ Foreign words:
# 'żmijewski', 'יגאל', 'כרמון'


# 👉 Dataset has:

# Different languages
# Names of actors/directors
# 🧠 IMPORTANT INSIGHT

# 👉 CountVectorizer does NO filtering by default
```
